```python
df = pd.read_excel(r"C:\Users\Nada\Downloads\Customer Call List (1).xlsx")
df
```

```python
df= df.drop_duplicates()
df
```

```python
df = df.drop(columns = "Not_Useful_Column")
df
```

```python
# remove unwanted characters (1,2,3,.,_,/) from the beginning and end of Last_Name
df["Last_Name"] = df["Last_Name"].str.strip("123._/")
```

```python
df
```

```python
df["Phone_Number"] = df["Phone_Number"].str.replace('nan--','')
df["Phone_Number"] = df["Phone_Number"].str.replace('Na--','')
df
```

```python
# split Address column into Street_Address, State, and Zip_Code using comma as delimiter
df[["Street_Address", "State", "Zip_Code"]] = df["Address"].str.split(',',n=2, expand=True)
df
```

```python
df["Do_Not_Contact"] = df["Do_Not_Contact"].str.replace('Yes','Y')
df["Do_Not_Contact"] = df["Do_Not_Contact"].str.replace('No','N')
df
```

```python
# replace all missing (NaN) values in the DataFrame with empty strings
df = df.fillna('')
df = df.replace('N/a','')
```

```python
df
```

```python
# remove rows where Do_Not_Contact is 'Y'
for x in df.index:
    if df.loc[x, "Do_Not_Contact"] == 'Y':
        df.drop(x, inplace=True)

df
#df = df[df["Do_Not_Contact"] != 'Y']
```

```python
for x in df.index:
    if df.loc[x, "Phone_Number"] == '':
        df.drop(x, inplace=True)

df

#Another way to drop null values
#df = df.dropna(subset="Phone_Number"), inplace=True)
```

```python
# reset index after deleting rows and drop old index
df = df.reset_index(drop=True)
df
```
