```python
cafe['Total Spent'] = pd.to_numeric(cafe['Total Spent'], errors='coerce')
```

```python
cafe['Total Spent'] = cafe['Total Spent'].fillna(cafe['Quantity']*cafe['Price Per Unit'])
```

### Step 8: Logical Validation

##### create a new column `Expected_Total` by multiplying `Quantity * Price Per Unit`

```python
cafe['Expected_Total'] = cafe['Quantity'] * cafe['Price Per Unit']
```

##### filter rows where Total Spent is NOT equal to Expected_Total

```python
cafe[cafe['Total Spent'] != cafe['Expected_Total']]
```

##### fill missing or incorrect Total Spent values using the Expected_Total calculation.

```python
cafe['Total Spent'] = cafe['Total Spent'].fillna(cafe['Expected_Total'])
```

```python
cafe.loc[(cafe['Total Spent'].isna()) | (cafe['Total Spent'] != cafe['Expected_Total']),'Total Spent'] = cafe['Expected_Total']
```
