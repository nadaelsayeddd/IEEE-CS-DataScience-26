### Step 1. Import the necessary libraries:

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
%reload_ext autoreload
%autoreload 2
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/07_Visualization/Tips/tips.csv). 

### Step 3. Assign it to a variable called tips

```python
tips = pd.read_csv(r"https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/07_Visualization/Tips/tips.csv")
```

### Step 4. Delete the Unnamed 0 column

```python
tips.drop("Unnamed: 0", axis=1, inplace=True)
```

### Step 5. Plot the total_bill column histogram

```python
plt.figure(figsize=(8,6))
sns.histplot(tips['total_bill'], bins=20, kde=False)
plt.title("Histogram of Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")
plt.show()
```

### Step 6. Create a scatter plot presenting the relationship between total_bill and tip

```python
plt.figure(figsize=(8,6))
sns.scatterplot(x='total_bill', y='tip', data=tips)
plt.title("Total Bill vs Tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()
```

### Step 7.  Create one image with the relationship of total_bill, tip and size.
#### Hint: It is just one function.

```python
sns.relplot(x='total_bill',y='tip',size='size',data=tips)
```

### Step 8. Present the relationship between days and total_bill value

```python
plt.figure(figsize=(8,6))
sns.boxplot(x='day', y='total_bill', data=tips)
plt.title("Total Bill Distribution Across Days")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.show()
```

### Step 9. Create a scatter plot with the day as the y-axis and tip as the x-axis, differ the dots by sex

```python
plt.figure(figsize=(8,8))
sns.scatterplot(x='tip', y='day', hue='sex', data=tips)
plt.title("Tip vs Day (colored by Sex)")
plt.xlabel("Tip")
plt.ylabel("Day")
plt.show()
```

### Step 10.  Create a box plot presenting the total_bill per day differetiation the time (Dinner or Lunch)

```python
plt.figure(figsize=(8,6))
sns.boxplot(x='day', y='total_bill', hue='time', data=tips)
plt.title("Total Bill per Day (Lunch vs Dinner)")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.show()
```

### Step 11. Create two histograms of the tip value based for Dinner and Lunch. They must be side by side.

```python
plt.figure(figsize=(6,4))
plt.subplot(1,2,1)
sns.histplot(tips[tips['time'] == 'Lunch']['tip'], bins=15, kde=False)
plt.title("Lunch Tips")
plt.xlabel("Tip")
plt.subplot(1,2,2)
sns.histplot(tips[tips['time'] == 'Dinner']['tip'], bins=15, kde=False)
plt.title("Dinner Tips")
plt.xlabel("Tip")
plt.tight_layout()
plt.show()
```

### Step 12. Create two scatterplots graphs, one for Male and another for Female, presenting the total_bill value and tip relationship, differing by smoker or no smoker
### They must be side by side.

```python
plt.subplot(1,2,1)
sns.scatterplot(x='total_bill',y='tip',hue='smoker',data=tips[tips['sex'] == 'Male'])
plt.title("Male: Total Bill vs Tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.subplot(1,2,2)
sns.scatterplot(x='total_bill',y='tip',hue='smoker',data=tips[tips['sex'] == 'Female'])
plt.title("Female: Total Bill vs Tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.tight_layout()
plt.show()
```
